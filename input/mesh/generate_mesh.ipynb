```python
import subprocess
import time
```

```python
# Plate dimensions
L, W = 0.05, 0.05
lpz = 0.007555556       # Process zone length (based on ceramic material properties)
lc = L / 100       # Mesh characteristic length
h = 0.002

p = 2               # Polynomial order
```

```python
# Genrate the .geo file

output_name = f"plate_{L:.2e}x{W:.2e}x{h:.2e}_P{p}"
 
geo_file = f"""
// Define the points
Point(1) = {{{-L/2}, {-W/2}, {-h/2}, {lc}}};
Point(2) = {{{L/2}, {-W/2}, {-h/2}, {lc}}};
Point(3) = {{{L/2}, {W/2}, {-h/2}, {lc}}};
Point(4) = {{{-L/2}, {W/2}, {-h/2}, {lc}}};
Point(5) = {{{-L/2}, {-W/2}, {h/2}, {lc}}};
Point(6) = {{{L/2}, {-W/2}, {h/2}, {lc}}};
Point(7) = {{{L/2}, {W/2}, {h/2}, {lc}}};
Point(8) = {{{-L/2}, {W/2}, {h/2}, {lc}}};

// Define the bottom face lines
Line(1) = {{1, 2}};
Line(2) = {{2, 3}};
Line(3) = {{3, 4}};
Line(4) = {{4, 1}};

// Define the top face lines
Line(5) = {{5, 6}};
Line(6) = {{6, 7}};
Line(7) = {{7, 8}};
Line(8) = {{8, 5}};

// Define vertical lines connecting top and bottom faces
Line(9) = {{1, 5}};
Line(10) = {{2, 6}};
Line(11) = {{3, 7}};
Line(12) = {{4, 8}};

// Define the line loops for surfaces
Line Loop(1) = {{1, 2, 3, 4}};  // Bottom face
Line Loop(2) = {{5, 6, 7, 8}};  // Top face
Line Loop(3) = {{1, 10, -5, -9}}; // Side 1
Line Loop(4) = {{2, 11, -6, -10}}; // Side 2
Line Loop(5) = {{3, 12, -7, -11}}; // Side 3
Line Loop(6) = {{4, 9, -8, -12}}; // Side 4

// Define surfaces
Plane Surface(1) = {{1}}; // Bottom face
Plane Surface(2) = {{2}}; // Top face
Plane Surface(3) = {{3}}; // Side 1
Plane Surface(4) = {{4}}; // Side 2
Plane Surface(5) = {{5}}; // Side 3
Plane Surface(6) = {{6}}; // Side 4

// Define the volume
Surface Loop(1) = {{1, 2, 3, 4, 5, 6}};
Volume(1) = {{1}};

// Physical groups
Physical Surface("bottom") = {{1}};
Physical Surface("top") = {{2}};
Physical Surface("sides") = {{3, 4, 5, 6}};
Physical Volume("plate") = {{1}};

// Meshing properties
Mesh.Algorithm = 5;
Mesh.ElementOrder = {p};
"""

with open(output_name + ".geo", "w") as f:
    f.write(geo_file)
```

```python
# Generate the mesh
ret = subprocess.Popen("gmsh -3 " + output_name + ".geo -o " + output_name + ".msh", shell=True)
if ret.returncode:
    print("Beware, gmsh could not run: mesh is not regenerated")
else:
    print("Mesh generated")
```
