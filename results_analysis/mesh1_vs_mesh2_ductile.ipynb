```python
import h5py
```

```python
mesh_1_file = h5py.File('/scratch/saporito/Frag3D/mesh_1_L_over_50/data.h5', 'r')
mesh_2_file = h5py.File('/scratch/saporito/Frag3D/mesh_2_L_over_100/data.h5', 'r')
```

```python
print(mesh_1_file.keys())
print(mesh_2_file.keys())
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sorted_keys_m1 = sorted(mesh_1_file.keys(), key=lambda x: int(x.split('_')[-1]))
sorted_keys_m2 = sorted(mesh_2_file.keys(), key=lambda x: int(x.split('_')[-1]))
```

```python
print("Sorted keys for Mesh 1:", sorted_keys_m1)
print("Sorted keys for Mesh 2:", sorted_keys_m2)
```

```python
# Savding variation of ['econ', 'edis', 'ekin', 'epot', 'erev', 'nb_fragments', 'time', 'total_energy', 'work']
# wrt their value in step_0 at each step
econ_m1 = []
edis_m1 = []
ekin_m1 = []
epot_m1 = []
erev_m1 = []
nb_fragments_m1 = []
total_energy_m1 = []
for step in sorted_keys_m1:
    step_data = mesh_1_file[step]
    econ_m1.append(step_data.attrs['econ'])
    edis_m1.append(step_data.attrs['edis'])
    ekin_m1.append(step_data.attrs['ekin'])
    epot_m1.append(step_data.attrs['epot'])
    erev_m1.append(step_data.attrs['erev'])
    nb_fragments_m1.append(step_data.attrs['nb_fragments'])
    total_energy_m1.append(step_data.attrs['total_energy'])
steps_1 = np.array([int(k.split('_')[-1]) for k in sorted_keys_m1])
econ_m1 = np.array(econ_m1)
edis_m1 = np.array(edis_m1)
ekin_m1 = np.array(ekin_m1)
epot_m1 = np.array(epot_m1)
erev_m1 = np.array(erev_m1)
nb_fragments_m1 = np.array(nb_fragments_m1)
total_energy_m1 = np.array(total_energy_m1)
```

```python
econ_m2 = []
edis_m2 = []
ekin_m2 = []
epot_m2 = []
erev_m2 = []
nb_fragments_m2 = []
total_energy_m2 = []
for step in sorted_keys_m2:
    step_data = mesh_2_file[step]
    econ_m2.append(step_data.attrs['econ'])
    edis_m2.append(step_data.attrs['edis'])
    ekin_m2.append(step_data.attrs['ekin'])
    epot_m2.append(step_data.attrs['epot'])
    erev_m2.append(step_data.attrs['erev'])
    nb_fragments_m2.append(step_data.attrs['nb_fragments'])
    total_energy_m2.append(step_data.attrs['total_energy'])
steps_2 = np.array([int(k.split('_')[-1]) for k in sorted_keys_m2])
econ_m2 = np.array(econ_m2)
edis_m2 = np.array(edis_m2)
ekin_m2 = np.array(ekin_m2)
epot_m2 = np.array(epot_m2)
erev_m2 = np.array(erev_m2)
nb_fragments_m2 = np.array(nb_fragments_m2)
total_energy_m2 = np.array(total_energy_m2)
```

```python
# Plotting the variation of these quantities over time steps
plt.figure(figsize=(20, 15))
plt.subplot(3, 2, 1)
plt.plot(steps_1, econ_m1, label='Econ - Mesh 1')
plt.plot(steps_2, econ_m2, label='Econ - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Econ')
plt.title('Econ vs Step')
plt.subplot(3, 2, 2)
plt.plot(steps_1, edis_m1, label='Edis - Mesh 1')
plt.plot(steps_2, edis_m2, label='Edis - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Edis')
plt.title('Edis vs Step')
plt.subplot(3, 2, 3)
plt.plot(steps_1, ekin_m1, label='Ekin - Mesh 1')
plt.plot(steps_2, ekin_m2, label='Ekin - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Ekin')
plt.title('Ekin vs Step')
plt.subplot(3, 2, 4)
plt.plot(steps_1, epot_m1, label='Epot - Mesh 1')
plt.plot(steps_2, epot_m2, label='Epot - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Epot')
plt.title('Epot vs Step')
plt.subplot(3, 2, 5)
plt.plot(steps_1, erev_m1, label='Erev - Mesh 1')
plt.plot(steps_2, erev_m2, label='Erev - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Erev')
plt.title('Erev vs Step')
plt.subplot(3, 2, 6)
plt.plot(steps_1, total_energy_m1, label='Total Energy - Mesh 1')
plt.plot(steps_2, total_energy_m2, label='Total Energy - Mesh 2')
plt.xticks([])
plt.xlabel('Step')
plt.ylabel('Total Energy')
plt.title('Total Energy vs Step')
plt.show()
```

```python
# Let's do the plots of these quantities but for the time, not the time steps
# Let's save the times and map them to the quantities for both meshes
times_m1 = []
for step in sorted_keys_m1:
    step_data = mesh_1_file[step]
    times_m1.append(step_data.attrs['time'])
times_m1 = np.array(times_m1)
times_m2 = []
for step in sorted_keys_m2:
    step_data = mesh_2_file[step]
    times_m2.append(step_data.attrs['time'])
times_m2 = np.array(times_m2)
```

```python
# Plotting the variation of the quantities over time
plt.figure(figsize=(20, 15))
plt.subplot(3, 2, 1)
plt.plot(times_m1, econ_m1, label='Econ - Mesh 1')
plt.plot(times_m2, econ_m2, label='Econ - Mesh 2')
plt.xticks([])
plt.xlabel('Time')  
plt.ylabel('Econ')
plt.title('Econ vs Time')
plt.subplot(3, 2, 2)
plt.plot(times_m1, edis_m1, label='Edis - Mesh 1')
plt.plot(times_m2, edis_m2, label='Edis - Mesh 2')
plt.xticks([])
plt.xlabel('Time')
plt.ylabel('Edis')
plt.title('Edis vs Time')
plt.subplot(3, 2, 3)
plt.plot(times_m1, ekin_m1, label='Ekin - Mesh 1')
plt.plot(times_m2, ekin_m2, label='Ekin - Mesh 2')
plt.xticks([])
plt.xlabel('Time')
plt.ylabel('Ekin')
plt.title('Ekin vs Time')
plt.subplot(3, 2, 4)
plt.plot(times_m1, epot_m1, label='Epot - Mesh 1')
plt.plot(times_m2, epot_m2, label='Epot - Mesh 2')
plt.xticks([])
plt.xlabel('Time')
plt.ylabel('Epot')
plt.title('Epot vs Time')
plt.subplot(3, 2, 5)
plt.plot(times_m1, erev_m1, label='Erev - Mesh 1')
plt.plot(times_m2, erev_m2, label='Erev - Mesh 2')
plt.xticks([])
plt.xlabel('Time')
plt.ylabel('Erev')
plt.title('Erev vs Time')
plt.subplot(3, 2, 6)
plt.plot(times_m1, total_energy_m1, label='Total Energy - Mesh 1')
plt.plot(times_m2, total_energy_m2, label='Total Energy - Mesh 2')
plt.xticks([])
plt.xlabel('Time')
plt.ylabel('Total Energy')
plt.title('Total Energy vs Time')
plt.show()
```

```python
# Save the variation of each quantity with respect to its value at step_0 for each mesh
initial_econ_m1 = econ_m1[0]
initial_edis_m1 = edis_m1[0]
initial_ekin_m1 = ekin_m1[0]
initial_epot_m1 = epot_m1[0]
initial_erev_m1 = erev_m1[0]
diff_econ_1 = []
diff_edis_1 = []
diff_ekin_1 = []
diff_epot_1 = []
diff_erev_1 = []
diff_total_energy_1 = []
initial_total_energy_m1 = total_energy_m1[0]
for step in sorted_keys_m1:
    step_data = mesh_1_file[step]
    diff_econ_1.append(step_data.attrs['econ'] - initial_econ_m1)
    diff_edis_1.append(step_data.attrs['edis'] - initial_edis_m1)
    diff_ekin_1.append(step_data.attrs['ekin'] - initial_ekin_m1)
    diff_epot_1.append(step_data.attrs['epot'] - initial_epot_m1)
    diff_erev_1.append(step_data.attrs['erev'] - initial_erev_m1)
    diff_total_energy_1.append(step_data.attrs['total_energy'] - initial_total_energy_m1)
```

```python
initial_econ_m2 = econ_m2[0]
initial_edis_m2 = edis_m2[0]
initial_ekin_m2 = ekin_m2[0]
initial_epot_m2 = epot_m2[0]
initial_erev_m2 = erev_m2[0]
initial_total_energy_m2 = total_energy_m2[0]
diff_econ_2 = []
diff_edis_2 = []
diff_ekin_2 = []
diff_epot_2 = []
diff_erev_2 = []
diff_total_energy_2 = []
for step in sorted_keys_m2:
    step_data = mesh_2_file[step]
    diff_econ_2.append(step_data.attrs['econ'] - initial_econ_m2)
    diff_edis_2.append(step_data.attrs['edis'] - initial_edis_m2)
    diff_ekin_2.append(step_data.attrs['ekin'] - initial_ekin_m2)
    diff_epot_2.append(step_data.attrs['epot'] - initial_epot_m2)
    diff_erev_2.append(step_data.attrs['erev'] - initial_erev_m2)
    diff_total_energy_2.append(step_data.attrs['total_energy'] - initial_total_energy_m2)
```

```python
# Plot the variation of each quantity with respect to its value at step_0 for each mesh over time
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(12, 7))
plt.plot(times_m1, diff_econ_1, label='Econ', linewidth=2, alpha=0.8)
plt.plot(times_m1, diff_edis_1, label='Edis', linewidth=2, alpha=0.8)
plt.plot(times_m1, diff_ekin_1, label='Ekin', linewidth=2, alpha=0.8)
plt.plot(times_m1, diff_epot_1, label='Epot', linewidth=2, alpha=0.8)
plt.plot(times_m1, diff_erev_1, label='Erev', linewidth=2, alpha=0.8)
plt.plot(times_m1, diff_total_energy_1, label='Total Energy', 
         color='black', linestyle='--', linewidth=2.5, zorder=10)
plt.title('Energy Variation Relative to Step 0 (Mesh 1)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Time', fontsize=12)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), frameon=True, fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(12, 7))
plt.plot(times_m2, diff_econ_2, label='Econ', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_edis_2, label='Edis', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_ekin_2, label='Ekin', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_epot_2, label='Epot', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_erev_2, label='Erev', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_total_energy_2, label='Total Energy', 
         color='black', linestyle='--', linewidth=2.5, zorder=10)
plt.title('Energy Variation Relative to Step 0 (Mesh 2)', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Time', fontsize=12)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), frameon=True, fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
# Plotting the variation of just the total energy wrt step_0 for both meshes over time
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(12, 7))
plt.plot(times_m1, diff_edis_1, label='Dissipated Energy - Mesh 1', color='blue', linewidth=2, alpha=0.8)
plt.plot(times_m2, diff_edis_2, label='Dissipated Energy - Mesh 2', color='orange', linewidth=2, alpha=0.8)
plt.title('Dissipated Energy Variation Relative to Step 0', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Time', fontsize=12)
plt.ylabel('Dissipated Energy Variation', fontsize=12)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), frameon=True, fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
```
